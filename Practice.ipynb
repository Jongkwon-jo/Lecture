```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 3. PDF 문서 로드
loader = PyPDFLoader(r"C:\Users\User\Downloads\여비규정.pdf")
pages = loader.load()

# 4. 텍스트 분할 설정 (Chunking)
text_splitter = RecursiveCharacterTextSplitter(
                                                chunk_size=500, # 하나의 청크 크기(500자는 대략 한글 기준으로 3~5개 문장을 포함하므로, 특정 질문에 대한 핵심 답변을 담기에 충분한 크기)
                                                chunk_overlap=100, # 문맥 단절 방지용 중복 구간(보통 전체 청크 크기의 10%~20%를 오버랩으로 설정)
                                                length_function=len
                                                )

# 5. 분할 실행
docs = text_splitter.split_documents(pages)

# 6. 결과 확인
print(f"총 {len(pages)}페이지에서 {len(docs)}개의 청크 생성됨")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

# 2. 벡터 DB 생성 (임베딩 모델 설정)
# 한국어 성능이 좋은 모델을 사용합니다.
embeddings = HuggingFaceEmbeddings(model_name="jhgan/ko-sroberta-multitask")
vector_db = FAISS.from_documents(docs, embeddings)

print(f"Vector DB 구축 완료: {len(docs)}개의 청크 저장됨")
```

```python
#Step 3 RAG 시스템 구축(RAG 프롬프트 & 실행 코드)

import openai

# 3. 답변 생성 함수
def generate_rag_response(query):
    # (1) 검색: 질문과 가장 유사한 상위 3개 청크 추출
    search_results = vector_db.similarity_search(query, k=3)
    
    # 검색된 내용 합치기 및 출처 정보 생성
    context_text = "\n\n".join([doc.page_content for doc in search_results])
    sources = [f"{doc.metadata.get('source', 'Unknown')}: {doc.metadata.get('page', '0')}p" for doc in search_results]
    
    # (2) LLM 호출
    client = openai.OpenAI()
    
    system_prompt = """당신은 '인사팀 AI 매니저'입니다. 
                        제공된 [Context]를 바탕으로 질문에 답하세요. 정보가 없으면 "정보 없음"이라고 하세요.
                        답변이 제공된 경우, 답변 끝에 출처를 명시하세요.
                        정보가 없으면 출처를 명시하지 마세요."""

    response = client.chat.completions.create(
        model="gpt-4o",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": f"[Context]\n{context_text}\n\n[Question]\n{query}"}
        ],
        temperature=0.2,
    )
    
    return response.choices[0].message.content, sources
```

```python
# --- [Step 4] 실행 (간이 검색 포함) ---
# 질문 예시
query = "국외 출장 시 식비는 어떻게 지급되나요?"

# 답변 생성 함수 실행
answer, source_list = generate_rag_response(query)

print("-" * 50)
print(f"[질문]: {query}")
print(f"\n[AI 답변]:\n{answer}")
print(f"\n[출처 확인]: {set(source_list)}")
print("-" * 50)
```

```python
#Step 5: 리포팅 자동화(최종 보고서 생성 & 코드 구현)

import datetime

def create_markdown_report(qa_results):
    """
    Q&A 결과 리스트를 받아 마크다운 형식의 보고서 파일로 저장하는 함수
    """
    # 1. 보고서 헤더 생성
    today = datetime.date.today()
    md_text = f"# 📑 신입사원 FAQ 보고서\n"
    md_text += f"**생성일:** {today}\n**작성자:** 인사팀 AI 매니저\n\n"
    md_text += "---\n\n## 1. 질의응답 (Q&A)\n\n"

    # 2. Q&A 리스트 순회 및 포맷팅
    # enumerate의 시작 번호를 1로 설정하여 가독성 확보
    for idx, item in enumerate(qa_results, 1):
        question = item['question']
        answer = item['answer']
        # .get()을 사용해 키가 없을 경우를 대비한 기본값 설정
        source = item.get('source', '근거 문서를 찾을 수 없음')
        
        # Markdown 형식 작성
        md_text += f"### Q{idx}. {question}\n"
        md_text += f"**A:** {answer}\n\n"
        md_text += f"> **📌 근거:** {source}\n\n"
        md_text += "---\n\n"

    # 3. 하단 푸터 추가
    md_text += f"\n*본 보고서는 RAG(Retrieval-Augmented Generation) 시스템에 의해 자동 생성되었습니다.*"

    # 4. 파일로 저장
    filename = f"faq_report_{today.strftime('%Y%m%d')}.md" # 날짜를 포함한 파일명
    with open(filename, "w", encoding="utf-8") as f:
        f.write(md_text)
    
    print(f"✅ 보고서가 성공적으로 생성되었습니다: {filename}")
    return filename
```

```python
import os

# 1. 질문 리스트 준비 (여비규정에서 신입사원이 가장 궁금해할 법한 내용)
queries = [
    "국외 출장 시 일비 지급 기준은 어떻게 되나요?",
    "시내 출장 시 교통비 지원 범위를 알려주세요.",
    "숙박비 영수증이 없어도 정액 지급이 가능한가요?",
    "자가용을 이용해 출장을 갈 경우 운임 계산 방식은?",
    "출장 중 식비는 일괄 지급인가요, 실비 정산인가요?"
]

# 2. 결과를 담을 리스트
qa_results = []

print("🚀 RAG 기반 질의응답 및 리포팅 시작...")

# 3. 루프를 돌며 각 질문에 대한 답변 추출 및 저장
for i, query in enumerate(queries, 1):
    print(f"[{i}/{len(queries)}] 질문 처리 중: {query}")
    
    try:
        # Step 4에서 만든 함수 호출 (답변과 출처를 리턴)
        answer, source_list = generate_rag_response(query)
        
        # 결과 저장
        qa_results.append({
            "question": query,
            "answer": answer,
            "source": ", ".join(set(source_list)) # 중복 제거된 출처
        })
    except Exception as e:
        print(f"⚠️ 질문 '{query}' 처리 중 오류 발생: {e}")

# 4. 최종 보고서 생성 (Step 5 함수 호출)
if qa_results:
    report_file = create_markdown_report(qa_results)
    print(f"\n✨ 모든 작업이 완료되었습니다!")
    print(f"최종 보고서가 '{report_file}'로 저장되었습니다.")
else:
    print("❌ 생성된 답변이 없어 보고서를 만들지 못했습니다.")
```
